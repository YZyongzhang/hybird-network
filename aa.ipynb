```python
import os
import pickle
path = './dataset/pickle/v2/greedy(hybirdnetwork_with_RGBD_train)'
scenes = [os.path.join(path , i ) for i in os.listdir(path)]
files = []
for s in scenes:
    files.extend([os.path.join(s , i ) for i in os.listdir(s)])
```

```python
for file in files:
    print(file)
    with open(file , 'rb') as f:
        data = pickle.load(f)
    level_diseance_to_goal_json.setdefault(data['info'][-1]['na'] , []).append(file)
```
